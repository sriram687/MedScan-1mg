## Section 4️⃣: Dependency Analysis & Data Flow

### 🔗 Key Technology Relationships

#### **Authentication Flow**
```
User Login
  ├─ Clerk (Frontend) ─────→ Clerk Backend
  │   └─ JWT Token Generation
  │
  └─ Express Backend
      ├─ JWT Verification
      ├─ bcryptjs for Password Hashing
      └─ MongoDB for User Storage
```

#### **Payment Processing**
```
Checkout Page (React Stripe Elements)
  ├─ Stripe Integration
  │   └─ Card Payment Processing
  └─ Razorpay Integration
      └─ UPI/NetBanking/Cards (Indians)

Payment Confirmation
  └─ Express Backend Updates Order Status
      └─ MongoDB Order Update
      └─ Nodemailer Sends Confirmation Email
```

#### **Prescription OCR Pipeline**
```
User Uploads Prescription Image
  ├─ Client: Tesseract.js (Fallback)
  └─ Server: TrOCR / Donut Transformer
      ├─ Image Processing
      ├─ Text Extraction
      ├─ Medicine Name Fuzzy Matching
      └─ Cloudinary Storage + Order Link
```

#### **Data Pipeline** 
```
Puppeteer Web Scraper
  ├─ Scrapes 1mg.com (3,028 medicines)
  ├─ Extracts: Name, Price, Composition, Uses, etc.
  └─ MongoDB Aggregation & Indexing
      └─ Full-Text Search Capabilities
```
